 implement the apriori algo to find frequent item sets,
 generate the frequent items with given min support=50%
 generate association rules from frequent itemsets
 calculate confidence,support and lift
 t1=milk,bread
 t2=bread,butter
 t3=milk,bread,butter
 t4=milk,butter

```python
transactions = [
    ["milk","bread"],
    ["bread","butter"],
    ["milk","bread","butter"],
    ["milk","butter"]
]

min_support_count = 2

def count_support(itemset):
    count = 0
    for t in transactions:
        found = True
        for item in itemset:
            if item not in t:
                found = False
                break
        if found:
            count += 1
    return count

items = ["milk","bread","butter"]

print("Frequent Itemsets:")
for item in items:  #checking for 1 item
    sup = count_support([item])
    if sup >= min_support_count:
        print("{",item,"} support =",sup)

pairs = []
for i in range(len(items)):   #checking for 2 items
    for j in range(i+1,len(items)):
        pair = [items[i],items[j]]
        sup = count_support(pair)
        if sup >= min_support_count:
            print("{",items[i],",",items[j],"} support =",sup)
            pairs.append(pair)
```

```python
print("Association Rules-")
for pair in pairs:
    sup_pair = count_support(pair)
    for item in pair:
        other = pair[0] if pair[1]==item else pair[1]
        sup_item = count_support([item])
        confidence = sup_pair / sup_item
        lift = confidence / (count_support([other]) / len(transactions))
        print(item,"->",other,"\tSupport =", (sup_pair/len(transactions))*100,"%","\tConfidence =", round(confidence*100,2),"%","\tLift =", round(lift,2))
```
