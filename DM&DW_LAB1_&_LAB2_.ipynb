```python
#write a python function that takes a list of intervals as input where each interval is presented as pair of integers and returns a new list of merged intervals
intervals=[]
n=int(input("Enter number of intervals"))
for i in range(n):
    s,e=map(int,input("Enter start and end of interval").split())
    intervals.append([s,e])

intervals.sort()
merged=[]
for interval in intervals:
        if not merged or merged[-1][1]<interval[0]:
            merged.append(interval)
        else:
            merged[-1][1]=max(merged[-1][1],interval[1])
print("Output:",merged)
print()
```

```python
#takes a sorted list and target integer as input and performs binary search to find the target,return the index of the target, -1 if not present
arr=list(map(int,input("Enter sorted list").split()))
target=int(input("Enter target"))
l=0
r=len(arr)-1
index=-1
while l<=r:
  mid=(l+r)//2
  if arr[mid]==target:
    index=mid
    break
  elif arr[mid]<target:
    l=mid+1
  else:
    r=mid-1
print("Output:",index)
print()
```

```python
#takes a 2D matrix as a list of lists and returns the transpose of the matrix
rows=int(input("Enter number of rows"))
cols=int(input("Enter number of columns"))
matrix=[]
for i in range(rows):
    row=list(map(int,input("Enter a row").split()))
    matrix.append(row)

transpose=[]
for j in range(cols):
  row=[]
  for i in range(rows):
    row.append(matrix[i][j])
  transpose.append(row)

print("Output:",transpose)
print()
```

```python
#ques 2.1
import numpy as np

data_list = [2, 4, 6, 8, 10, 12, 14, 16]
numpy_array = np.array(data_list)
print("Original NumPy array:")
print(numpy_array)

mean_val = np.mean(numpy_array)
median_val = np.median(numpy_array)
std_dev_val = np.std(numpy_array)
print("\nMean:", mean_val)
print("Median:", median_val)
print("Standard Deviation:", std_dev_val)

reshaped_array = numpy_array.reshape(2, 4)
print("\nReshaped 2x4 array:")
print(reshaped_array)

square_root_array = np.sqrt(numpy_array)
print("\nElement-wise square root array:")
print(square_root_array)

max_index = np.argmax(numpy_array)
print("\nIndex of maximum element:", max_index)

elements_greater_than_8 = numpy_array[numpy_array > 8]
print("\nElements greater than 8:")
print(elements_greater_than_8)
```

```python
import pandas as pd

data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eve'],
    'Age': [28, 22, 35, 30, 25],
    'Department': ['HR', 'Engineering', 'HR', 'Engineering', 'Finance'],
    'Salary': [50000, 60000, 55000, 62000, 45000]
}
df = pd.DataFrame(data)
print(df)
df.info()

average_salary = df['Salary'].mean()
print("Average Salary:", average_salary)

engineering_max_age = df[df['Department'] == 'Engineering']['Age'].max()
print(engineering_max_age)

high_earners = df[df['Salary'] > 50000]['Name']
print("Employees with Salary > 50000:\n", high_earners.to_string(index=False))

new_employee = pd.DataFrame([{'Name': 'Frank', 'Age': 40, 'Department': 'Finance', 'Salary': 55000}])
df = pd.concat( [df/, new_employee], ignore_index=True)
print("adding Frank:\n", df)

df = df[df['Name'] != 'Eve']
print("removing Eve:\n", df)
```
